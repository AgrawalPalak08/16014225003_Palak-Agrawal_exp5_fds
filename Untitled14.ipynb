```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

uploaded = files.upload()
df = pd.read_csv('grib_data.csv')

df = df[['latitude', 'longitude', 't2m', 'skt', 'sst']]

print(df.head())
print("\nMissing values:")
print(df.isnull().sum())
```

```python

# Select rows where both values are available
data = df[['skt', 'sst']].dropna()

x = data['skt'].values
y = data['sst'].values

# Split data into training and testing sets
np.random.seed(42)
indices = np.random.permutation(len(x))
split = int(0.8 * len(x))

train = indices[:split]
test = indices[split:]

x_train = x[train]
y_train = y[train]

# Calculate regression coefficients
w1 = np.sum((x_train - x_train.mean()) *
            (y_train - y_train.mean())) / np.sum((x_train - x_train.mean())**2)

w0 = y_train.mean() - w1 * x_train.mean()

# Predict test values
pred = w0 + w1 * x[test]

# Calculate accuracy
mse = np.mean((y[test] - pred)**2)
rmse = np.sqrt(mse)

print("Simple Linear Regression")
print("Equation: sst =", w0, "+", w1, "* skt")
print("MSE:", mse)
print("RMSE:", rmse)

# Scatter plot
plt.scatter(x[test][:500], y[test][:500], label='Actual')
plt.scatter(x[test][:500], pred[:500], label='Predicted')
plt.xlabel('Skin Temperature (skt)')
plt.ylabel('Sea Surface Temperature (sst)')
plt.title('Simple Linear Regression')
plt.legend()
plt.show()
```

```python

# Select complete rows for training and testing
data = df.dropna()

X = data[['latitude', 'longitude', 't2m', 'skt']].values
y = data['sst'].values

# Split data
np.random.seed(42)
indices = np.random.permutation(len(y))
split = int(0.8 * len(y))

train = indices[:split]
test = indices[split:]

X_train = X[train]
X_test = X[test]
y_train = y[train]
y_test = y[test]

# Add intercept column
X_train = np.column_stack((np.ones(len(X_train)), X_train))
X_test = np.column_stack((np.ones(len(X_test)), X_test))

# Calculate coefficients using least squares
b = np.linalg.pinv(X_train.T @ X_train) @ X_train.T @ y_train

# Predict values
pred_multi = X_test @ b

# Calculate accuracy
mse_multi = np.mean((y_test - pred_multi)**2)
rmse_multi = np.sqrt(mse_multi)

print("Multiple Linear Regression")
print("Coefficients:", b)
print("MSE:", mse_multi)
print("RMSE:", rmse_multi)
```

```python

comparison = pd.DataFrame({
    'Model': ['Simple Linear Regression',
              'Multiple Linear Regression'],
    'MSE': [mse, mse_multi],
    'RMSE': [rmse, rmse_multi]
})

print(comparison)
```

```python

# Train using rows where sst is available
known = df.dropna()

X_known = known[['latitude', 'longitude', 't2m', 'skt']].values
y_known = known['sst'].values

X_known = np.column_stack((np.ones(len(X_known)), X_known))

b_final = np.linalg.pinv(X_known.T @ X_known) @ X_known.T @ y_known

# Find rows where sst is missing
missing = df['sst'].isnull()

X_missing = df.loc[missing,
                   ['latitude', 'longitude', 't2m', 'skt']].values

X_missing = np.column_stack((np.ones(len(X_missing)), X_missing))

# Fill missing values
df.loc[missing, 'sst'] = X_missing @ b_final

print("Missing sst values after prediction:")
print(df['sst'].isnull().sum())

print(df.head())
```
